```python
from qiskit import QuantumCircuit
from IPython.display import display
import matplotlib.pyplot as plt
```

state preparation

```python
# (a) Prepare |0>
qc = QuantumCircuit(1)
qc.id(0)

display(qc.draw(output='mpl', initial_state=True,
                filename='Fig_3_1a.png'))


# (b) Prepare |1>
qc = QuantumCircuit(1)
qc.x(0)

display(qc.draw(output='mpl', initial_state=True,
                filename='Fig_3_1b.png'))


# (c) Prepare |+>
qc = QuantumCircuit(1)
qc.h(0)

display(qc.draw(output='mpl', initial_state=True,
                filename='Fig_3_1c.png'))


# (d) Prepare |->
qc = QuantumCircuit(1)
qc.x(0)
qc.h(0)

display(qc.draw(output='mpl', initial_state=True,
                filename='Fig_3_1d.png'))
```

Bell-state preparation

```python
qc = QuantumCircuit(2)

# Prepare superposition
qc.h(0)

# Entangle the two qubits
qc.cx(0, 1)

display(qc.draw(output='mpl', initial_state=True,
                filename='Fig_3_2.png'))
```

Bit-flip encoding

```python
qc = QuantumCircuit(3)

# Encode the input qubit
qc.cx(0, 1)
qc.cx(0, 2)

display(qc.draw(output='mpl',
                filename='Fig_5_1.png'))
```

Bit-flip syndrome extraction

```python
qc = QuantumCircuit(5, 2)

# First check: Z1 Z2
qc.cx(0, 3)
qc.cx(1, 3)

# Second check: Z1 Z3
qc.cx(0, 4)
qc.cx(2, 4)

qc.barrier()

# Measure the ancillas
qc.measure(3, 0)
qc.measure(4, 1)

display(qc.draw(output='mpl', cregbundle=False,
                filename='Fig_5_2.png'))
```

Phase-flip encoding

```python
qc = QuantumCircuit(3)

# Repetition-code encoding
qc.cx(0, 1)
qc.cx(0, 2)

qc.barrier()

# Change to the Hadamard basis
qc.h([0, 1, 2])

display(qc.draw(output='mpl',
                filename='Fig_5_3.png'))
```

Phase flip syndrome extraction

```python
qc = QuantumCircuit(5, 2)

# Change the data-qubit basis
qc.h([0, 1, 2])

qc.barrier()

# Extract the two parities
qc.cx(0, 3)
qc.cx(1, 3)

qc.cx(0, 4)
qc.cx(2, 4)

qc.barrier()

# Return to the phase-flip-code basis
qc.h([0, 1, 2])

# Measure the ancillas
qc.measure(3, 0)
qc.measure(4, 1)

display(qc.draw(output='mpl', cregbundle=False,
                filename='Fig_5_4.png'))
```
